```python
from glob import glob
import os
from vampires_dpp.organization import header_table
```

# HD114710 - Making a subtable and saving sub file list with vampires_dpp.table

```python
data_file_path = "/home/shared/exoserver/VAMPIRES/20240218_fourth_download/"
files = glob(data_file_path + "VMPA*.fits.fz")
files = [f for f in files if os.path.isfile(f)]
# NOTE: To do this, we need to 
table = header_table(files)
```

```python
HD114710_subtable = table.query("OBJECT == 'HD114710'")
print(HD114710_subtable)
```

```python
paths_str = "\n".join(str(p) for p in HD114710_subtable["path"])
with open(data_file_path + "HD114710_filelist.txt", "w") as fh:
    fh.write(paths_str); fh.write("\n")
```
